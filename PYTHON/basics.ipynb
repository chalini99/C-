```python
#creating class
class student: # __init__ is a constructor method that is automatically called when an object of the class is created. It initializes the object's attributes with the provided values.
    def __init__(self, name, age): #self- current object, allowing each object to store and access its own data
        self.name = name # self.name and self.age are instance attributes, unique to each 'student' object created from the class.
        self.age= age  #object - is an instance of a class, holds its own data.
student1 = student('chalini', 21)
print(student1.name)
print(student1.age)
```

```python
## Reverse a String WITHOUT slicing
def rev(s):
    result = ""
    for i in range(len(s) -1, -1, -1):
        result+=s[i]
    return result
```

```python
s = "hello"
print(rev(s))
```

```python
## reverse a string with slicing
s = "hello"
reverse = s[::-1]
print(reverse)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
```

```python
class Dog(Animal):

    def sound(self):
        print("Bark")
```

```python
dog = Dog()
dog.sound()
```

```python
# Static Method
#A static method doesn't need information from either the object or class.
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b
print(Calculator.add(10, 20))
```

```python
# positional arguments - are passed to a function in the order they are defined in the function signature. The first argument corresponds to the first parameter, the second argument to the second parameter, and so on.
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

s = Student("Chalini", 21)


# Keyword arguments - are passed to a function by explicitly specifying the parameter names along with their corresponding values. This allows you to provide arguments in any order, as long as you specify the parameter names.
s = Student(name="Chalini", age=21)
```

```python
# *args in OOP - collects positional arguments into a tuple
class addition:

    def add(self, *args):
        return sum(args)
c = addition()
print(c.add(10, 20))
print(c.add(10, 20, 30, 40))
```

```python
## **kwargs in OOP - collects keyword arguments into a dictionary
class Student:

    def display(self, **kwargs):
        print(kwargs)
s = Student()
s.display(name="Chalini", age=21, branch="AIML")
```

```python
# One OOP example that combines everything

from abc import ABC, abstractmethod

class Animal(ABC):

    count = 0

    def __init__(self, name):
        self.name = name
        Animal.count += 1

    @abstractmethod
    def sound(self):
        pass

    @classmethod
    def total_animals(cls):
        return cls.count

    @staticmethod
    def info():
        print("Animals are living organisms.")

    @property
    def animal_name(self):
        return self.name


class Dog(Animal):

    def sound(self):
        return "Bark"


dog = Dog("Bruno")

print(dog.name)
print(dog.sound())
print(dog.animal_name)

Animal.info()

print(Animal.total_animals())
```

```python
# decorators - where a function wraps another function to modify its behaviour without changing its souce code

def decorator(func):
    def wrapper():
        print('start')
        func()
        print('end')
    return wrapper
@decorator
def hello():
    print("Hello")
hello()
```
